Recommended order:

1. **01_Foundations.ipynb** — tensors, autograd, shapes, devices, memory, views, broadcasting, randomness  
2. **02_Training_and_Data.ipynb** — Dataset/DataLoader, training loops, losses, optimizers, schedulers, AMP  
3. **03_nn_Module_DeepDive.ipynb** — nn.Module, parameters/buffers, hooks, init, normalization, regularization  
4. **04_Vision_TorchVision.ipynb** — transforms, CNNs, transfer learning, detection/segmentation patterns  
5. **05_Audio_TorchAudio.ipynb** — waveforms, spectrograms, audio models, augmentation patterns  
6. **06_NLP_Tokenizers_and_TextPipelines.ipynb** — vocab, padding/masks, word/char/BPE tokenizers, datasets  
7. **07_Transformers_From_Scratch.ipynb** — attention, causal masks, decoder-only LM, generation, KV-cache concept  
8. **08_HuggingFace_Transformers.ipynb** — AutoTokenizer/AutoModel, Trainer, Accelerate, fine-tuning, generation/chat  
9. **09_Performance_Compile_Export_Profiling.ipynb** — torch.compile, torch.export, profiler, memory practices  
10. **10_Distributed_DDP_FSDP.ipynb** — DDP scripts, samplers, gradient sync, FSDP sharding patterns  
11. **11_TorchFX_TorchFunc_CustomOps.ipynb** — FX graphs, transforms, torch.func (vmap/grad), extension overview  
12. **12_Quantization_and_Deployment.ipynb** — PTQ/QAT overview, dynamic quantization, serving patterns
