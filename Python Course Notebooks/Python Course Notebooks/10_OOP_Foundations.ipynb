## 1.
L1: Classes and instances

- `__init__` initializes a new instance.
- `self` is the instance receiving the method call.

````python

class Point:
    def __init__(self, x: float, y: float):
        self.x = x
        self.y = y

    def move(self, dx: float, dy: float) -> None:
        self.x += dx
        self.y += dy

p = Point(1, 2)
p.move(0.5, -1)
print(p.x, p.y)
````

## 2.
L2: Representation and debugging (`__repr__`)

A good `__repr__` makes debugging dramatically easier.

````python

class User:
    def __init__(self, username: str):
        self.username = username

    def __repr__(self) -> str:
        return f"User(username={self.username!r})"

print(User("alice"))
````

## 3.
L3: Class methods and static methods

- `@classmethod` receives `cls` (the class); often used for alternate constructors.
- `@staticmethod` is namespaced function (no `self`/`cls`).

````python

class Temperature:
    def __init__(self, c: float):
        self.c = c

    @classmethod
    def from_f(cls, f: float) -> "Temperature":
        return cls((f - 32) * 5/9)

    @staticmethod
    def c_to_f(c: float) -> float:
        return c * 9/5 + 32

t = Temperature.from_f(98.6)
print(t.c)
print(Temperature.c_to_f(0))
````

## 4.
L4: Inheritance vs composition

- Inheritance expresses “is-a”.
- Composition expresses “has-a”.

Prefer composition unless inheritance clearly models your domain.

````python

class Engine:
    def start(self) -> str:
        return "vroom"

class Car:
    def __init__(self, engine: Engine):
        self.engine = engine

    def drive(self) -> str:
        return self.engine.start() + " 🚗"

print(Car(Engine()).drive())
````

## 5.
L5: Dataclasses (standard library)

Dataclasses remove boilerplate for “data containers”.

They can generate:
- `__init__`
- `__repr__`
- `__eq__`
- ordering, frozen immutability, default factories, etc.

````python

from dataclasses import dataclass, field

@dataclass
class Config:
    host: str = "localhost"
    port: int = 8080
    tags: list[str] = field(default_factory=list)

cfg = Config(tags=["dev"])
print(cfg)
````

## 7.
L7: Dataclasses advanced options

- `frozen=True` for immutability
- `order=True` for comparisons
- `slots=True` (3.10+) for memory savings

````python

from dataclasses import dataclass

@dataclass(frozen=True, order=True, slots=True)
class Version:
    major: int
    minor: int
    patch: int

print(Version(1,2,0) < Version(1,10,0))
````

## 8.
L8: Enums (stdlib `enum`)

Enums represent fixed sets of symbolic values.

````python

from enum import Enum, auto

class Status(Enum):
    PENDING = auto()
    RUNNING = auto()
    DONE = auto()

s = Status.PENDING
print(s, s.name, s.value)
````
