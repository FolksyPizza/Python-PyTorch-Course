## 1.
L1: Calling C with `ctypes` (stdlib)

`ctypes` can call functions from shared libraries (DLL/.so/.dylib).
This is powerful, but you must manage:
- types
- memory ownership
- platform differences

````python

import ctypes, sys

def try_strlen() -> int | None:
    try:
        if sys.platform.startswith("win"):
            lib = ctypes.CDLL("msvcrt.dll")
        elif sys.platform == "darwin":
            lib = ctypes.CDLL("libc.dylib")
        else:
            lib = ctypes.CDLL("libc.so.6")
        lib.strlen.argtypes = [ctypes.c_char_p]
        lib.strlen.restype = ctypes.c_size_t
        return int(lib.strlen(b"hello"))
    except Exception as e:
        print("ctypes demo failed on this platform:", e)
        return None

print("strlen('hello'):", try_strlen())
````

## 2.
L2: Subprocess vs native bindings

- `subprocess` is easiest and most portable (treats other programs like black boxes).
- native bindings (ctypes/cffi/pybind11) offer speed and direct access, but add build complexity.

## 3.
L3: Writing extension modules (overview)

Options:
- **Cython**: Python-like syntax → C extension
- **pybind11**: modern C++ bindings
- **cffi**: write C declarations; build bindings

Use these when profiling proves Python is the bottleneck.
