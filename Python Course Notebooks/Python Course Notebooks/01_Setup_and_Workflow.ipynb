## 1.
L1: Running Python (REPL, scripts, modules)

Ways to run code:

- Interactive REPL: `python`
- Run a script file: `python my_script.py`
- Run a module: `python -m package.module`
- Run a package: `python -m package` (requires `package/__main__.py`)
- Jupyter notebooks: good for exploration; ship production code as packages/modules.

### Why `python -m` matters

`python -m ...` runs the target **as a module** using your current interpreter/venv.
That avoids “wrong Python” problems and makes imports behave like they will in production.

````python

# Try it: the __name__ switch
def main() -> None:
    print("Hello from main()")

print("__name__ is:", __name__)
if __name__ == "__main__":
    main()
````

## 2.
L2: Virtual environments (venv) and dependency isolation

Rules of thumb:

- One project → one environment.
- Never `pip install` into your system Python.
- Pin dependencies (exact versions) for reproducibility.

### Requirements files vs pyproject

- `requirements.txt` is common (works everywhere).
- `pyproject.toml` is the modern standard for building/packaging and can also manage dependencies.
- Pick **one** workflow per project to reduce confusion.

````python

# Try it: programmatically locate site-packages for the current interpreter
import site, sys
print("executable:", sys.executable)
for p in site.getsitepackages():
    print("site-packages:", p)
````
