## 1.
L1: Web services

Two main server interfaces:
- WSGI (sync): Flask, Django
- ASGI (async): FastAPI, Starlette, Django (async parts)

In production you run an app via a server like `gunicorn` (WSGI) or `uvicorn` (ASGI).

### Minimal WSGI app (stdlib `wsgiref`)

````python

from wsgiref.simple_server import make_server

def app(environ, start_response):
    body = b"hello wsgi"
    start_response("200 OK", [("Content-Type", "text/plain"), ("Content-Length", str(len(body)))])
    return [body]

# In a script you could run:
# with make_server("127.0.0.1", 8000, app) as httpd:
#     httpd.serve_forever()

print("Defined a WSGI app (run it in a script to serve).")
````
