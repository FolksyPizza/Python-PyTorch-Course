## 1.
L1: Names, objects, and assignment

In Python, variables are **names** bound to **objects**.

- Assignment binds a name to an object.
- Multiple names can refer to the same object.
- “Mutability” determines whether an object can be changed in place.

````python

a = [1, 2, 3]
b = a          # b points to the same list object
b.append(4)

print("a:", a)  # changed!
print("b:", b)

# identity vs equality
print("a is b:", a is b)   # same object
print("a == b:", a == b)   # same contents
````

## 2.
L2: Core scalar types

- `int` (arbitrary precision)
- `float` (IEEE 754 double)
- `bool` (`True`/`False`, subclass of `int`)
- `None` (singleton: `None`)

Also: `complex`, `bytes`, and `str`.

````python

x = 10
y = 3.5
z = None
flag = True

print(type(x), type(y), type(z), type(flag))
print(True + True + False)  # bools behave like ints (1/0)
````

## 3.
L3: Operators and precedence (what to remember)

Common operators:
- arithmetic: `+ - * / // % **`
- comparisons: `== != < <= > >=`
- boolean: `and or not`
- membership: `in`, `not in`
- identity: `is`, `is not`

Rule: use `is` only for **singletons** like `None` (and sometimes `True/False`).

````python

value = None
if value is None:
    print("value is None (good)")

# Beware: `== None` can be overridden by user types.
````

## 4.
L4: Conversions, truthiness, and common pitfalls

- `int("12")`, `float("3.14")`, `str(123)`
- Empty containers are falsy: `[]`, `{}`, `set()`, `""`
- `0` and `0.0` are falsy; non-zero numbers are truthy

````python

examples = [0, 1, "", "hi", [], [1], None, {}, {"a": 1}]
for e in examples:
    print(repr(e).ljust(10), "truthy?" , bool(e))
````

## 5.
L5: Numeric precision (decimal vs float)

`float` is fast but not exact for many decimal fractions.
Use `decimal.Decimal` for currency/financial calculations.

````python

from decimal import Decimal, getcontext

print(0.1 + 0.2)                 # floating-point surprise
getcontext().prec = 28
print(Decimal("0.1") + Decimal("0.2"))
````

## 7.
L7: Introspection basics (`type`, `id`, `dir`, `help`)

Python is *inspectable* at runtime.

- `type(x)` → the class/type
- `id(x)` → identity (implementation detail; in CPython it's the memory address)
- `dir(x)` → attribute names
- `help(x)` → doc/help (interactive)

````python

x = {"a": 1}
print("type:", type(x))
print("id  :", id(x))
print("has attr 'keys'?", hasattr(x, "keys"))
print("first 5 attrs:", dir(x)[:5])
````

## 8.
L8: `bytes`, `bytearray`, and `memoryview`

- `bytes` is immutable binary data.
- `bytearray` is mutable binary data.
- `memoryview` provides a zero-copy view into binary buffers (useful for performance).

````python

b = b"abc"
ba = bytearray(b)
ba[0] = ord("z")
print(b, ba)

mv = memoryview(ba)
mv[1] = ord("y")
print("bytearray after memoryview edit:", ba)
````

## 9.
L9: Comparisons: chained comparisons and `math.isclose`

Chained comparisons are real:

```python
0 <= x < 10
```

For floats, use `math.isclose` instead of `==` for computed values.

````python

import math

x = 0.1 + 0.2
print("x:", x)
print("x == 0.3:", x == 0.3)
print("isclose:", math.isclose(x, 0.3, rel_tol=1e-9, abs_tol=0.0))
````
