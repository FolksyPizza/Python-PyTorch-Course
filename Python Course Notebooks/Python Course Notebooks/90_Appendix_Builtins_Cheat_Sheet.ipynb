```python

xs = [3, 1, 2]
print(sorted(xs))
print(list(enumerate(xs, start=1)))
print(list(zip([1,2], ["a","b"])))
print(any(x > 2 for x in xs), all(x > 0 for x in xs))
```

## Idioms

- prefer `with open(...)` for files
- prefer `pathlib.Path` for paths
- use `dict.get(key, default)` for safe lookups
- use `set` for membership tests
- prefer `python -m ...` to ensure correct interpreter
- prefer small pure functions and clear boundaries
