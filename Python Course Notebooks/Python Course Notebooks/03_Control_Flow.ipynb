## 1.
L1: if / elif / else and short-circuit logic

`and` / `or` short-circuit (they stop evaluating as soon as the result is known).

Common patterns:
- guard clauses (return early)
- ternary: `x if cond else y`

````python

def classify(x: int) -> str:
    if x < 0:
        return "negative"
    elif x == 0:
        return "zero"
    return "positive"

for n in [-2, 0, 5]:
    print(n, "->", classify(n))
````

## 2.
L2: for-loops, range, enumerate, zip

Python’s `for` iterates over *iterables* (not just numeric ranges).

````python

items = ["a", "b", "c"]
for i, item in enumerate(items, start=1):
    print(i, item)

xs = [1, 2, 3]
ys = [10, 20, 30]
for x, y in zip(xs, ys):
    print(x, y, "sum:", x + y)
````

## 3.
L3: while-loops, break/continue, and loop-else

`else` on loops runs **only if the loop did not break**.
This is underused and extremely useful for search-style loops.

````python

nums = [2, 4, 6, 7, 8]
target_is_prime = True

for n in nums:
    if n % 2 == 0 and n != 2:
        continue
    # naive prime test for demo
    if n > 2 and n % 2 == 0:
        target_is_prime = False
        break
else:
    # runs only if loop didn't break
    print("loop completed without break")

print("target_is_prime:", target_is_prime)
````

## 4.
L4: Structural pattern matching (`match` / `case`) (Python 3.10+)

Use it when you’d otherwise write long `if/elif` chains, especially on structured data.

````python

def handle(event: dict) -> str:
    match event:
        case {"type": "click", "x": int(x), "y": int(y)}:
            return f"click at ({x}, {y})"
        case {"type": "key", "key": str(k)}:
            return f"key {k}"
        case _:
            return "unknown"

print(handle({"type": "click", "x": 3, "y": 9}))
print(handle({"type": "key", "key": "Enter"}))
print(handle({"type": "noop"}))
````

## 5.

## 6.
L6: Exercises

1. Write `fizzbuzz(n)` returning a list of strings from 1..n.
2. Write a search loop over a list that uses `for ... else` to report “not found”.
3. Use `match/case` to parse simple commands like `{"cmd":"add","a":1,"b":2}`.

````python

def fizzbuzz(n: int) -> list[str]:
    out: list[str] = []
    for i in range(1, n + 1):
        if i % 15 == 0:
            out.append("FizzBuzz")
        elif i % 3 == 0:
            out.append("Fizz")
        elif i % 5 == 0:
            out.append("Buzz")
        else:
            out.append(str(i))
    return out

assert fizzbuzz(5) == ["1","2","Fizz","4","Buzz"]
assert fizzbuzz(15)[-1] == "FizzBuzz"
print("ok")
````

## 7.
L7: Comprehensions vs loops (readability guideline)

Use comprehensions for:
- mapping/filtering where the intent is obvious

Use loops for:
- complex multi-step logic
- side effects (logging, I/O)

````python

# mapping
xs = [1,2,3,4,5]
ys = [x*x for x in xs if x % 2 == 1]
print(ys)
````

## 8.
L8: The walrus operator (`:=`) (Python 3.8+)

Use it when it improves clarity by avoiding repeated work.
Don’t use it to be clever.

````python

data = ["  ", "hello", "", "world"]
clean = []
for s in data:
    if (t := s.strip()):
        clean.append(t)
print(clean)
````

## 9.
L9: “Early return” as a control-flow tool

Deep nesting reduces readability. Prefer guard clauses.

````python

def parse_positive_int(s: str) -> int:
    s = s.strip()
    if not s:
        raise ValueError("empty")
    n = int(s)
    if n <= 0:
        raise ValueError("must be positive")
    return n

print(parse_positive_int(" 12 "))
````
