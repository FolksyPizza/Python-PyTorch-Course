## Glossary (selected)

- **object**: a runtime value with identity, type, and state
- **name**: a binding to an object (what many languages call a variable)
- **mutable / immutable**: whether state can change in-place
- **iterable / iterator**: produces values; iterator has `__next__`
- **generator**: iterator produced by a function with `yield`
- **context manager**: object usable with `with`, providing cleanup
- **duck typing**: behavior matters, not nominal type
- **GIL**: CPython lock that limits parallel bytecode execution in threads
- **idempotent**: safe to repeat without changing result (important for retries)
- **serialization**: convert objects ↔ bytes/text for storage/transport
- **wheel**: built distribution artifact (`.whl`) for fast installs
- **sdist**: source distribution (`.tar.gz`) that must be built on install

## Mental models

### “Everything is an object”
Even functions, classes, and modules are objects.

### “Names point to objects”
Assignment doesn’t copy; it binds names.

### “Types are about behavior”
Use protocols/ABCs and small interfaces.

### “Measure performance”
Use timeit/cProfile before “optimizing”.
