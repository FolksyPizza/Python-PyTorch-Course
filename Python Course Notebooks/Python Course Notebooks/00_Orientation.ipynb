## 1.
L1: What “knowing Python” means

Python proficiency is a stack:

- **Syntax fluency:** you can read/write basic code without friction.
- **Semantic accuracy:** you understand execution order, mutability, scope, exceptions, and the object model.
- **Standard library mastery:** you know what already exists before reinventing it.
- **Engineering discipline:** you can structure projects, test, debug, type-check, lint, and package code.
- **Performance + reliability:** you can profile, optimize, and avoid foot‑guns (security, concurrency, I/O).
- **Ecosystem literacy:** you can pick the right third‑party tools and integrate them safely.

## 4.
L4: Suggested “starter kit” tools (what pros actually use)

- **ruff** (fast linter) + **black** (formatter)
- **pytest** (testing)
- **mypy** or **pyright** (static typing)
- **pre-commit** (automate formatting/linting in Git hooks)
- **uv** or **pip-tools** / **poetry** (dependency management) — choose one per project
- **logging** (stdlib) for diagnostics (avoid `print()` in production)
