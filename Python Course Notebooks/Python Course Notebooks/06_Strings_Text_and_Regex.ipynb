## 1.
L1: Unicode and encoding (the rules)

- `str` is Unicode text.
- `bytes` is raw 8-bit data.
- Encoding converts `str → bytes` (e.g., UTF‑8).
- Decoding converts `bytes → str`.

````python

text = "café ☕"
b = text.encode("utf-8")
roundtrip = b.decode("utf-8")
print(text, b, roundtrip)
````

## 2.
L2: String operations: split/join, strip, replace

Rule: build strings with `''.join(...)` (especially in loops).

````python

line = "  a,b,c  "
parts = line.strip().split(",")
print(parts)
print("|".join(parts))
print("hello".replace("l", "L"))
````

## 3.
L3: f-strings and format specifiers

Format mini-language is powerful:

- `f"{x:.2f}"` fixed decimals
- `f"{n:,}"` thousand separators
- `f"{pct:.1%}"` percent formatting

````python

x = 3.14159
n = 1234567
pct = 0.1234
print(f"{x:.2f}", f"{n:,}", f"{pct:.1%}")
````

## 4.
L4: Regex essentials (and when *not* to use regex)

Regex is great for **token-ish** patterns; it’s not great for nested structures.
Always start with simple string methods; use regex when it’s clearly simpler.

````python

import re

pattern = re.compile(r"(?P<user>[a-z0-9_]+)@(?P<domain>[a-z0-9.-]+)\.(?P<tld>[a-z]+)", re.I)
m = pattern.search("Contact: alice_123@example.co.uk")
print(m.group(0))
print(m.groupdict())
````

## 5.
L5: Safer parsing patterns

- Use `str.partition`, `str.split`, `csv`, `json`, `urllib.parse` when possible.
- Regex should be your last step, not your first.

````python

from urllib.parse import urlparse, parse_qs

u = "https://example.com/search?q=python&sort=desc&q=typing"
p = urlparse(u)
qs = parse_qs(p.query)
print(p.scheme, p.netloc, p.path)
print(qs)
````

## 6.
L6: Exercises

1. Write `normalize_whitespace(s)` that collapses runs of whitespace into single spaces.
2. Extract all emails from a text block (use regex).
3. Write `slugify(title)` using only stdlib: lowercase, replace spaces with hyphens, drop non-alphanumerics.

````python

import re

_ws = re.compile(r"\s+")

def normalize_whitespace(s: str) -> str:
    return _ws.sub(" ", s).strip()

assert normalize_whitespace("  a\t b\nc   ") == "a b c"
print("ok")
````

## 7.
L7: Encoding errors and robust decoding

When decoding bytes from unknown sources, you may need an error strategy:
- `errors="strict"` (default)
- `errors="replace"`
- `errors="ignore"` (danger: silent data loss)

````python

bad = b"\xff\xfe\xfd"
print(bad.decode("utf-8", errors="replace"))
````

## 8.
L8: Regex flags and performance

Compile regexes you reuse.
Common flags:
- `re.I` ignore case
- `re.M` multiline (^/$ match line boundaries)
- `re.S` dot matches newline

````python

import re
text = "A\nb\nC"
print(re.findall(r"^[a-z]$", text, flags=re.I | re.M))
````

## 9.
L9: `str.translate` for fast character mapping

Useful for stripping punctuation or normalizing.

````python

import string

table = str.maketrans("", "", string.punctuation)
s = "Hello, world!!!"
print(s.translate(table))
````
