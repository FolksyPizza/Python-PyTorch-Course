## 1.
L1: The modern baseline: `pyproject.toml`

`pyproject.toml` is the central configuration file for:
- building packages
- tool configuration (ruff, black, mypy, pytest, etc.)

A minimal example:

```toml
[project]
name = "my_project"
version = "0.1.0"
requires-python = ">=3.11"
dependencies = ["requests>=2.31"]

[project.scripts]
my-cli = "my_project.cli:main"
```

## 2.
L2: Editable installs (`pip install -e .`)

During development, install your project in editable mode:

```bash
python -m pip install -e .
```

Then `import my_project` works from anywhere, reflecting your local changes.

## 3.
L3: Building wheels and sdists

```bash
python -m build
```

Outputs:
- source distribution (`.tar.gz`)
- wheel (`.whl`) — preferred install artifact

## 4.
L4: Versioning

Use semantic-ish versioning:
- MAJOR: breaking
- MINOR: new features
- PATCH: bug fixes

Automate versions with tags or tools (e.g., `setuptools-scm`) if desired.

## 5.
L5: Entry points and CLIs

If you publish a CLI, use `project.scripts` (shown above).
In code, keep CLI thin: parse args, call library functions.

## 7.
L7: Dependency pinning and lockfiles

For applications (not libraries), pin exact versions:

- `pip-tools` → `requirements.txt` + `requirements.lock`
- `poetry` → `poetry.lock`
- `uv` → lock files / fast installs (tooling varies by workflow)

For libraries, specify ranges and test against multiple versions.

## 8.
L8: Extras and optional dependencies

Extras let users install optional features:

```toml
[project.optional-dependencies]
dev = ["pytest", "ruff", "mypy"]
```

Install:
```bash
pip install .[dev]
```
